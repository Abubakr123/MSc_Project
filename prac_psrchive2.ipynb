```python
arch = psrchive.Archive_load('B0138+59_D20170507T091201_122_133.ar')
print 'the name of the source. \n', arch.get_source()

data = arch.get_data()
print 'The dimensions of the data array which are; subintegration, polarization, channel, and profile bin \n',data.shape

duration = arch.get_Integration(0).get_duration()
print 'the obs durtion \n', duration

nchan = arch.get_Integration(0).get_nchan()
print 'The Number of frewuency channels \n' , nchan

subint=arch[0]
print 'The number of subintegration \n', subint

(b_mean, b_var) = subint.baseline_stats()
print 'The baseline stats and cal_levels \n',  b_mean.shape

#gitting the number of pulse profils  in the integration object
profile = subint.get_Profile(0,0)
profile_data = profile.get_amps()
print 'the single pulse profil data \n ' , profile_data.shape

mjd_dbl = arch.get_Integration(0).get_start_time().in_days()
print 'The MJD class \n' , mjd_dbl

#producing the file image
#arch.bscrunch_to_nbin(1024)
#arch.fscrunch_to_nchan(488)
#arch.convert_state('Stokes')#arch.pscrunch()
arch.dedisperse()
arch.remove_baseline()
freq_lo = arch.get_centre_frequency() - arch.get_bandwidth()/2.0
freq_hi = arch.get_centre_frequency() + arch.get_bandwidth()/2.0
plt.imshow(data[:,0,:,:].mean(0),aspect = ('auto')) #extent=(0,1,freq_lo,freq_hi))#tscrunch()
plt.xlabel('Pulse phase')
plt.ylabel('Frequency (MHz)')
plt.show()
#pylab.savefig('python_plot_example.png')

```
